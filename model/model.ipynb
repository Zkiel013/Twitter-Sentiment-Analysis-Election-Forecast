```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('vader_lexicon')
```

```python
def preprocess_tweet(text):
    text = str(text).lower()
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#', '', text)
    text = re.sub(r'[^a-z0-9\s]', '', text)
    tokenizer = TweetTokenizer()
    tokens = tokenizer.tokenize(text)
    lemmatizer = WordNetLemmatizer()
    sw = set(stopwords.words('english'))
    clean_tokens = [lemmatizer.lemmatize(t) for t in tokens if t not in sw and t.isalpha()]
    return " ".join(clean_tokens)
```

```python
sia = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    scores = sia.polarity_scores(text)
    compound = scores['compound']
    label = 'positive' if compound >= 0.05 else 'negative' if compound <= -0.05 else 'neutral'
    return {**scores, 'label': label}
```

```python
df = pd.read_csv('tweets.csv', quoting=3, encoding='utf-8', on_bad_lines='skip', engine='python')
df['clean_text'] = df['text'].apply(preprocess_tweet)
sentiment_scores = df['clean_text'].apply(analyze_sentiment).apply(pd.Series)
df = pd.concat([df, sentiment_scores], axis=1)
```

```python
df.to_csv('tweets_with_sentiment.csv', index=False)
print(df[['text', 'clean_text', 'compound', 'label']].head())
```
